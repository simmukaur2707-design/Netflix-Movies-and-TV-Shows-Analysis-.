## 2. Import Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## 3. Load Dataset

First upload `netflix_titles.csv` in the Files section on the left side of Google Colab.

```python
df = pd.read_csv('/content/netflix_titles.csv')
print("Dataset loaded successfully")
```

## 6. Missing Values

```python
df.isnull().sum()
```

## 7. Basic Data Analysis

```python
print("Movies and TV Shows:")
print(df['type'].value_counts())
```

```python
print("Top 10 Release Years:")
print(df['release_year'].value_counts().head(10))
```

```python
print("Top 10 Ratings:")
print(df['rating'].value_counts().head(10))
```

## 8. Maximum, Minimum and Average Release Year

```python
years = df['release_year'].dropna()

print("Maximum Year:", np.max(years))
print("Minimum Year:", np.min(years))
print("Average Year:", round(np.mean(years), 2))
```

## 9. Graphs

### Graph 1: Movies and TV Shows

```python
df['type'].value_counts().plot(kind='bar')
plt.title("Movies and TV Shows on Netflix")
plt.xlabel("Type")
plt.ylabel("Number of Titles")
plt.show()
```

### Graph 2: Top 10 Release Years

```python
df['release_year'].value_counts().head(10).sort_index().plot(kind='bar')
plt.title("Top 10 Release Years")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")
plt.show()
```

### Graph 3: Top 10 Countries

```python
df['country'].dropna().str.split(', ').str[0].value_counts().head(10).plot(kind='bar')
plt.title("Top 10 Countries")
plt.xlabel("Country")
plt.ylabel("Number of Titles")
plt.xticks(rotation=45)
plt.show()
```

### Graph 4: Top 10 Ratings

```python
df['rating'].value_counts().head(10).plot(kind='bar')
plt.title("Netflix Content Ratings")
plt.xlabel("Rating")
plt.ylabel("Number of Titles")
plt.xticks(rotation=45)
plt.show()
```

## 10. Observations

```python
print("1. The dataset contains", df.shape[0], "rows.")
print("2. The dataset contains", df.shape[1], "columns.")
print("3. The most common content type is", df['type'].value_counts().idxmax())
print("4. The most common release year is", df['release_year'].value_counts().idxmax())
print("5. The most common rating is", df['rating'].value_counts().idxmax())
```
